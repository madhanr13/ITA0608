```python
import csv

# Read CSV file
with open("trainingdata.csv") as csvFile:
    data = list(csv.reader(csvFile))

print("\nTraining Data:\n")
for row in data:
    print(row)

# Attributes and target
concepts = [row[:-1] for row in data[1:]]
target = [row[-1] for row in data[1:]]

# Initialize Specific Hypothesis
specific_h = concepts[0].copy()

# Initialize General Hypothesis
general_h = [["?" for i in range(len(specific_h))]
             for j in range(len(specific_h))]

print("\nInitial Specific Hypothesis:")
print(specific_h)

print("\nInitial General Hypothesis:")
for row in general_h:
    print(row)

# Candidate Elimination Algorithm
for i, h in enumerate(concepts):

    if target[i] == "Yes":

        # Generalize S
        for x in range(len(specific_h)):
            if h[x] != specific_h[x]:
                specific_h[x] = '?'
                general_h[x][x] = '?'

    else:

        # Specialize G
        for x in range(len(specific_h)):
            if h[x] != specific_h[x]:
                general_h[x][x] = specific_h[x]
            else:
                general_h[x][x] = '?'

# Remove overly general rows
general_h = [g for g in general_h if g != ['?'] * len(specific_h)]

print("\nFinal Specific Hypothesis:")
print(specific_h)

print("\nFinal General Hypothesis:")
for row in general_h:
    print(row)
```
