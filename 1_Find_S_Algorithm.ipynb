```python
import csv

# Training data
with open("trainingdata.csv") as csvFile:
    data = list(csv.reader(csvFile))

# Initialize hypothesis
hypothesis = ['Ø'] * (len(data[0]) - 1)

print("Initial Hypothesis:")
print(hypothesis)

# FIND-S Algorithm
for i, example in enumerate(data):
    if example[-1] == 'Yes':
        for j in range(len(hypothesis)):
            if hypothesis[j] == 'Ø':
                hypothesis[j] = example[j]
            elif hypothesis[j] != example[j]:
                hypothesis[j] = '?'

        print(f"\nAfter Positive Example {i+1}:")
        print(hypothesis)

print("\nFinal Hypothesis:")
print(hypothesis)
```
